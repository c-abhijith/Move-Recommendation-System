```python
movies = data[["id","title","original_language","overview","genre"]]
```

```python
movies
```

```python
movies = data[["id","title","original_language","overview","genre"]].copy()
```

```python
movies["tags"] = movies["overview"] + movies["genre"]

```

```python
movies.head()
```

```python
movie_data = movies.drop(columns=["overview","genre"])
```

```python
movie_data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=10000, stop_words='english')
```

```python
cv
```

```python
vector=cv.fit_transform(movie_data['tags'].values.astype('U')).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vector)
```

```python
similarity
```

```python
movie_data[movie_data['title']=="The Godfather"].index[0]
```

```python
distance = sorted(list(enumerate(similarity[2])), reverse=True, key=lambda vector:vector[1])
for i in distance[0:5]:
    print(movie_data.iloc[i[0]].title)
```

```python
def recommand(movies):
    index=movie_data[movie_data['title']==movies].index[0]
    distance = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda vector:vector[1])
    for i in distance[0:5]:
        print(movie_data.iloc[i[0]].title)
```

```python
recommand("Iron Man")
```

```python
import pickle
import numpy as np



def split_data(data, num_parts=10):
    chunk_size = len(data) // num_parts
    return [data[i*chunk_size : (i+1)*chunk_size] for i in range(num_parts - 1)] + [data[(num_parts-1)*chunk_size:]]

# Split and save
similarity_parts = split_data(similarity)

for i, part in enumerate(similarity_parts):
    with open(f"similarity_part{i+1}.pkl", "wb") as f:
        pickle.dump(part, f)
```
